## 🎤 Artist Analysis

First, I explored which artists appear most frequently in my current everyday playlist.

```python
artist_counts = music["artist"].value_counts()

artist_counts
```

```python
import matplotlib.pyplot as plt

artist_counts.plot(kind="bar")

plt.title("Artists in My Current Everyday Playlist")
plt.xlabel("Artist")
plt.ylabel("Number of Songs")
plt.xticks(rotation=75)
plt.tight_layout()

plt.show()
```
