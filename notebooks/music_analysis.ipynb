# 🎧 Music Listening Analytics

This project explores my current everyday music rotation using Python and Pandas.

My music taste covers many genres, from R&B to Celtic music and beyond. Recently, however, I have been listening mostly to female artists and vocalists, so I decided to use this current listening phase as the starting point for the project.

The goal is to explore patterns in the music I naturally return to, while learning Python for data analysis.

## Genre Analysis

Now that the dataset includes music metadata, I can explore which genres appear most often in my current everyday listening.

```python
genre_counts = music["genre"].value_counts()

genre_counts
```

```python
genre_counts.plot(kind="bar")

plt.title("Genres in My Current Everyday Listening")
plt.xlabel("Genre")
plt.ylabel("Number of Songs")
plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

### Key Finding

Electronic music is the most represented genre in my current everyday listening, with 8 of the 19 songs in the dataset.

Pop and R&B are also strongly represented, while Hip-Hop and Rock appear less frequently in this specific listening period.

This was interesting to me because R&B is one of my favorite genres overall, but this dataset shows that my current listening habits are leaning more toward electronic and experimental sounds.

Since this dataset represents my current rotation rather than my complete music taste, I see these results as a snapshot of one particular listening phase.

## Release Year Analysis

I also wanted to explore how old or recent the music in my current rotation is.

This helps me see whether my everyday listening is concentrated around recent releases or distributed across different periods.

```python
year_counts = music["release_year"].value_counts().sort_index()

year_counts
```

```python
year_counts.plot(kind="line", marker="o")

plt.title("Release Years in My Current Listening")
plt.xlabel("Release Year")
plt.ylabel("Number of Songs")
plt.grid(axis="y", alpha=0.3)
plt.tight_layout()

plt.show()
```

```python
recent_songs = music[music["release_year"] >= 2020]

recent_percentage = len(recent_songs) / len(music) * 100

print("Songs released since 2020:", len(recent_songs))
print("Total songs:", len(music))
print("Percentage released since 2020:", round(recent_percentage, 1), "%")
```

### Key Finding

10 of the 19 songs in my current rotation were released from 2020 onward, representing 52.6% of the dataset.

This shows that my current listening leans slightly toward recent releases, while still including music from several different periods.

The oldest track in this dataset was released in 1979, which also shows that my everyday listening is not limited to one generation of music.

Since this dataset represents only my current rotation, these results describe this particular listening period rather than my overall music taste.

## Artist Origin Analysis

My current playlist includes artists from several countries, so I wanted to explore where the artists represented in this listening period come from.

```python
origin_counts = music["artist_origin"].value_counts()

origin_counts
```

```python
origin_counts.sort_values().plot(kind="barh")

plt.title("Artist Origins in My Current Listening")
plt.xlabel("Number of Songs")
plt.ylabel("Artist Origin")
plt.tight_layout()

plt.show()
```

### Key Finding

Artists from the USA and UK are the most represented in my current listening, accounting for most of the songs in this dataset.

At the same time, the playlist includes artists from Finland, Denmark, Australia, Sweden and Canada, showing some geographic diversity beyond the two dominant countries.

This analysis counts songs rather than unique artists, so artists represented by multiple tracks have a greater influence on the results.
